```
#q_2_average.py
numbers=[]
Total=0
for i in range (10):
  n=int(input("Enter number:"))
  numbers.append(n)
  Total=Total+n#FIND TOTAL NUMBER

Average=Total/10
print("numbers:",numbers)
print("sum:",Total)
print("Average:",Average)
```

```
#q3_remove.py
number=list(map(int,input("Enter number integers:").split()))
unique=[]
for n in number:
  if n  not in unique:
       unique.append(n)#ADD NUMBER
print("original list:",numbers)
print("list after removing duplicate:",unique)
```
